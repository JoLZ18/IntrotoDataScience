## Part 1 Numpy Slicing

NumPy is Python library used for working arrays.

What is Slicing? 

Slicing means taking elements from given index( i.e starting postion) to another given index(i.e stop or end position)

We sliced by : [start:end] and can be by [start:end:step].

by default meaning no value starts begin at 0, end without value considered the length of that dimension, and step will count as 1 by default.





```python
# We will creating an array with shape(3,3) 3 dimension array. Randomizing the given value using np.random.randn()

grid = np.random.randn(3,3)

grid

```

Creating a Python code that will do the slice to reproduce as a result the required shape as follows:
> 1. (2, 2)
> 2. (3, )
> 3. (3, 2)
> 4. (1, 2)
> 5. (1, 3)

```python
# your code here
# A shape: (2 x 2)
print(grid[:2,:2].shape)

# B  (shape: (3 x 1)
print(grid[0, :].shape)

# C shape: (3 x 2)
print(grid[: , :2].shape)

# D shape: (1,2)
print(grid[:1 , :2].shape)

# E shape: (1,3)
print(grid[:1 , :].shape)
```

Assign a value or change value in an multi-dimension array by a slice expression


```python
grid[0:3, 0:1] = 1.00000000
print(grid)

#list[start:end:step] - step is optional if given that sequence will be sliced with that step value
listNum = ["One_Child_Family","Couples_Many_Children","Total_families","Single_Parent_Family","Couples_One_Child","Total_Couples","fAMILY"]
listNum[2:6:3] = ["x","Y"]
print(listNum)
```
